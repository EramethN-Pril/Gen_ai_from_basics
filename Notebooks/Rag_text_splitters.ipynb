We need langchain-text-splitters

first Text splitting

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=20,chunk_overlap=5)
```

```python
text="Hi dear ! my name is Prince and i Love to learn , read and sharp my knowledge"
```

```python
split_txt=splitter.split_text(text)
split_txt
```

Document splitting

```python
from langchain_community.document_loaders import TextLoader,PyPDFLoader
```

```python
txtload=TextLoader("../Data/iot.txt")
text=txtload.load()
len(text)
```

```python
chunk_text=splitter.split_documents(text)
chunk_text
```

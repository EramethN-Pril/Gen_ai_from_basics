```python
from dotenv import load_dotenv
load_dotenv
import os
```

```python
from langchain_groq import ChatGroq
```

```python
llm=ChatGroq(model="openai/gpt-oss-20b")
```

```python
from serpapi_search_tools import news_search,web_search
```

```python
from langchain.agents import create_agent
agent=create_agent(model=llm,tools=[news_search(),web_search()],system_prompt="You are an agent which takes information from the api and then tell the result")
```

```python
res=agent.invoke({"messages":[{"role":"user","content":"tell me top ai news"}]})
print(res["messages"][-1].content)
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
from serpapi_search_tools import news_search,web_search
```

```python
llm=ChatGroq(model="openai/gpt-oss-20b")
```

```python
agent=create_agent(
    model=llm,
    tools=[web_search(),news_search()],
    checkpointer=MemorySaver(),
    system_prompt="You are an agent with memory which search for information if not available"
)
```

```python
res=agent.invoke({"messages":[{"role":"user","content":"I would love to hear some stories about my name"}]},{"configurable":{"thread_id":"1"}})
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langgraph.graph import START,END,StateGraph
from pydantic import BaseModel,Field
```

```python
class Graphstate(BaseModel):
    message:str
    
```

```python
graph=StateGraph(Graphstate)
```

```python
def question(state:Graphstate):
    state.message="Are u sure babu ?"
    return state
```

```python
graph.add_node("question",question)
```

```python
graph.add_edge(START,"question")
```

```python
def reply(state:Graphstate):
    state.message=f"{state.message}!Yes babu"
    return state
```

```python
graph.add_node("reply",reply)
graph.add_edge("question","reply")
graph.add_edge("reply",END)
```

```python
finalgraph=graph.compile()
```

```python
res=finalgraph.invoke({"message":""})
```

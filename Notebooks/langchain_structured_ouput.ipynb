```python
from dotenv import load_dotenv
load_dotenv()
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
from pydantic import BaseModel
from pydantic import Field
from typing import List
```

```python
class Movies(BaseModel):
    name:str=Field(description="Name of the movie")
    year:int=Field(ge=2000,description="Year of release")
    rating:float=Field(description="Rating on imdb")
    avg_rating:float=Field(description="average rating across all platforms")
structured_llm=llm.with_structured_output(Movies)
```

```python
structured_llm.invoke("List 3 movies release after 2007")
```

```python
class LisMovies(BaseModel):
    movies:List[Movies]=Field(description="List of the movies")
structred_llm2=llm.with_structured_output(LisMovies)
structred_llm2.invoke("List 3 Movies released after 2007 ")
```

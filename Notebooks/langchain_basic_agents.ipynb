```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
from langchain_groq import ChatGroq
```

```python
llm=ChatGroq(model="openai/gpt-oss-20b")
```

```python
@tool
def divide_numbers(a:int ,b:int):
    """ 
     it will return division of two numbers
     Args:
        a : first number
        b : second number
    """
    return (float)(a/b)

```

```python
@tool 
def multiply_numbers(c:int,d:float):
    """
    it will give the multiplication value of the two numbers 
    Args:
        c : first number 
        d :second number
    """
    return c*d
multiply_numbers.invoke({"c": 4, "d": 4.9})
```

```python
from langchain.agents import create_agent
agent1=create_agent(
    model=llm,tools=[multiply_numbers,divide_numbers],
    system_prompt="You are an agents that uses tools for multiplication and division of numbres"
)
rex=agent1.invoke({"messages":[{"role":"user","content":"what is multiplication of 3 and 4.9"},{"role":"user","content":"what is division of 45 and 11"}]})
print(rex["messages"][-1].content)
```
